# P3(수정). 법 제11조 블록 — 2026.10.8 시행본으로 교체

## 과정 기록
| 단계 | 내용 |
|---|---|
| 1 | 기존 블록 `p0018:b01`, `p0018:b02`는 문장 중간에서 잘려 있었음 |
| 2 | 시행 기준 확인: **2026.10.8 시행본**(18쪽)이 현행. 제2호 삭제 표시(2026.10.20)는 **2027.10.21 시행 개정**에 속하므로 아직 적용되지 않음 (사용자 확인) |
| 3 | 18쪽 본문에서 ①과 ③의 문장을 그대로 옮김 (개정 이력 `<개정 …>`은 제외) |
| 4 | 조각별로 18쪽 원문에서 확인 |
| 5 | 두 블록만 교체. 나머지 레코드는 그대로 두고, 임시 파일에 쓴 뒤 원자적으로 바꿈 |

## 블록 계획
- `p0018:b01` 제11조 ① — 신고 대상과 의사의 보고·신고 (2026.10.8 시행본, 제2호 포함)
- `p0018:b02` 제11조 ③ — 의료기관의 장 신고 기준 (즉시 / 24시간 / 7일)
- 제11조 ②(확인기관 직원 보고)는 이번에 넣지 않음. 필요하면 `p0018:b03`으로 나중에 추가.

```python
import json, os
from preprocess.prevention_legal.pages import DOCS, excerpt_in_page

KEY = "law"
DOC_ID = DOCS[KEY][0]
BLOCKS = "../../data/interim/prevention_legal/blocks.jsonl"
TARGETS = [f"{DOC_ID}:p0018:b01", f"{DOC_ID}:p0018:b02"]

# 2026.10.8 시행본 18쪽 문장 (개정 이력 제외)
NEW = {
    TARGETS[0]: ("제11조(의사 등의 신고) ① 의사, 치과의사 또는 한의사는 다음 각 호의 어느 하나에 해당하는 사실"
                 "(제16조제6항에 따라 표본감시 대상이 되는 제4급감염병으로 인한 경우는 제외한다)이 있으면 "
                 "소속 의료기관의 장에게 보고하여야 하고, 해당 환자와 그 동거인에게 질병관리청장이 정하는 "
                 "감염 방지 방법 등을 지도하여야 한다. 다만, 의료기관에 소속되지 아니한 의사, 치과의사 또는 "
                 "한의사는 그 사실을 관할 보건소장에게 신고하여야 한다. "
                 "1. 감염병환자등을 진단하거나 그 사체를 검안(檢案)한 경우 "
                 "2. 예방접종 후 이상반응자를 진단하거나 그 사체를 검안한 경우 "
                 "3. 감염병환자등이 제1급감염병부터 제3급감염병까지에 해당하는 감염병으로 사망한 경우 "
                 "4. 감염병환자로 의심되는 사람이 감염병병원체 검사를 거부하는 경우"),
    TARGETS[1]: ("③ 제1항 및 제2항에 따라 보고를 받은 의료기관의 장 및 제16조의2에 따른 감염병병원체 확인기관의 장은 "
                 "제1급감염병의 경우에는 즉시, 제2급감염병 및 제3급감염병의 경우에는 24시간 이내에, "
                 "제4급감염병의 경우에는 7일 이내에 질병관리청장 또는 관할 보건소장에게 신고하여야 한다."),
}
SECTION = {
    TARGETS[0]: "제3장 신고 및 보고 · 제11조(의사 등의 신고) ①",
    TARGETS[1]: "제3장 신고 및 보고 · 제11조(의사 등의 신고) ③",
}
PIECES = {
    TARGETS[0]: ["이 있으면 소속 의료기관의 장에게 보고하여야 하고", "관할 보건소장에게 신고하여야 한다",
                 "예방접종 후 이상반응자를 진단하거나", "감염병환자로 의심되는 사람이 감염병병원체 검사를 거부하는 경우"],
    TARGETS[1]: ["제1항 및 제2항에 따라 보고를 받은 의료기관의 장", "제1급감염병의 경우에는 즉시",
                 "제4급감염병의 경우에는 7일 이내에", "질병관리청장 또는 관할 보건소장에게 신고하여야 한다"],
}
```

```python
# 1) 원문 확인: 18쪽에서 조각별로 찾는다 (실패하면 저장하지 않는다)
for cid in TARGETS:
    for piece in PIECES[cid]:
        ok = excerpt_in_page(piece, KEY, 18)
        print(("원문 확인 " if ok else "확인 필요 "), cid[-3:], "|", piece[:30])
        assert ok, f"18쪽에서 조각을 찾지 못했습니다: {piece}"
```

```python
# 2) 교체: 대상 두 줄만 바꾸고 나머지는 그대로 둔다
with open(BLOCKS, encoding="utf-8") as f:
    records = [json.loads(line) for line in f if line.strip()]
before = {r["chunk_id"]: r for r in records}
assert all(cid in before for cid in TARGETS), "교체 대상이 파일에 없습니다."

out = []
for r in records:
    cid = r["chunk_id"]
    if cid in NEW:
        out.append({
            "chunk_id": cid,
            "document_id": DOC_ID,
            "text": NEW[cid],
            "pdf_page": 18,
            "printed_page": None,
            "section": SECTION[cid],
            "source_note": "감염병예방법 2026.10.8 시행본 18쪽 (2027.10.21 개정 미적용)",
        })
    else:
        out.append(r)

ids = [r["chunk_id"] for r in out]
assert len(ids) == len(set(ids)), "chunk_id 가 중복됩니다."

# 임시 파일에 쓴 뒤 한 번에 바꾼다
tmp = BLOCKS + ".tmp"
with open(tmp, "w", encoding="utf-8") as f:
    for r in out:
        f.write(json.dumps(r, ensure_ascii=False) + chr(10))
os.replace(tmp, BLOCKS)

# 3) 검증: 대상 두 개만 바뀌었는지
after = {json.loads(l)["chunk_id"]: json.loads(l) for l in open(BLOCKS, encoding="utf-8") if l.strip()}
changed = [c for c in before if before[c] != after.get(c)]
print("전체 블록:", len(after), "| 바뀐 블록:", changed)
assert sorted(changed) == sorted(TARGETS), "예상과 다른 블록이 바뀌었습니다."
print("나머지", len(after) - len(TARGETS), "개는 그대로 ✅")
```
