## 1. 성홍열 신고 기준 (진단·신고기준 104쪽)
아래 셀은 진단·신고기준 표 추출본에서 **104쪽 성홍열 항목**만 꺼내 보여 줍니다.

```python
from pathlib import Path

RAW = Path("../../data/raw/prevention_legal/기타지침_법정감염병_진단신고기준_본문표.md")
md_text = RAW.read_text(encoding="utf-8")

start = md_text.index("## PDF 104페이지")
end = md_text.index("## PDF 105페이지")
section = md_text[start:end]

for line in section.splitlines():
    if line.startswith(("| 신고범위", "| 신고시기", "| 신고방법")):
        print(line[:220])
```

**읽는 법**
- `신고범위`: 환자, 의사환자 → 의사가 진단한 경우에 신고 대상이 됩니다.
- `신고시기`: 24시간 이내 → 법 제11조 ③과 같은 기한입니다.
- `신고방법`: 관할 보건소에 팩스 또는 방역통합정보시스템으로 신고합니다.

## 4. 공통 ID 형식 — 현황
스키마 v3.1 6절 형식을 따릅니다. 현재 저장된 ID를 확인합니다.

```python
import csv, json

rows = list(csv.DictReader(open("../../data/specs/source_manifest.csv", encoding="utf-8")))
print("문서 ID (source_manifest):")
for r in rows:
    print("  ", r["document_id"])

blocks = [json.loads(l) for l in open("../../data/interim/prevention_legal/blocks.jsonl", encoding="utf-8")]
print("\n블록 chunk_id (blocks.jsonl):")
for b in blocks:
    print("  ", b["chunk_id"])
```

## 5. 법령 ID 결정 미리보기 (파일은 바꾸지 않습니다)
현재 법령 ID는 `law_infectious_disease_act_21521`입니다. 스키마의 `{발행처}_{약칭}_{판본}` 규칙에서 **발행처를 어떻게 적을지**가 결정 사항입니다.

- **A안:** `moleg_infectious_disease_act_21521` (법제처 약칭)
- **B안:** `law_infectious_disease_act_21521` (현재 유지)

아래 셀은 **A안을 택하면 어떤 블록 ID가 바뀌는지** 보여 줍니다. 실제 파일은 그대로 둡니다.

```python
OLD = "law_infectious_disease_act_21521"
NEW = "moleg_infectious_disease_act_21521"   # A안 (아직 선택 전)

affected = [b["chunk_id"] for b in blocks if b["document_id"] == OLD]
print(f"A안으로 바꾸면 바뀌는 블록: {len(affected)}개")
for cid in affected:
    print("  ", cid, "->", cid.replace(OLD, NEW))
print("\n(manifest 의 법령 행도 document_id 를 함께 바꿔야 합니다.)")
```
