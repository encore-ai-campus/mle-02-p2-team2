- `p0006:b01` 제1급감염병 정의 (6쪽, 인쇄 쪽 없음)
- `p0006:b02` 제2급감염병 정의 (6쪽)
- `p0007:b01` 제3급감염병 정의 (7쪽)
- `p0019:b03` 제13조 ① 보건소장 보고 (19쪽, 개정 이력 제외)

법 본문은 법 제2조·제13조 자체를 근거로 쓴다. 신고 주체(누가 신고하는지)의 근거는 제11·12조 블록이다. 제11조 현행 확인은 별도 항목으로 남긴다.

```python
import json
from preprocess.prevention_legal.pages import DOCS, excerpt_in_page

KEY = "law"
DOC_ID = DOCS[KEY][0]
BLOCKS = "../../data/interim/prevention_legal/blocks.jsonl"
existing = [json.loads(l) for l in open(BLOCKS, encoding="utf-8")]
existing_ids = {b["chunk_id"] for b in existing}
print("기존 블록:", len(existing))
```

```python
# 블록 계획: (PDF 쪽, 블록 번호, 소제목, 블록 글, 확인 조각 목록)
PLAN = [
    (6, 1, "제2조(정의) 제2호 · 제1급감염병",
     "“제1급감염병”이란 생물테러감염병 또는 치명률이 높거나 집단 발생의 우려가 커서 발생 또는 유행 즉시 신고하여야 하고, 음압격리와 같은 높은 수준의 격리가 필요한 감염병으로서 다음 각 목의 감염병을 말한다. 다만, 갑작스러운 국내 유입 또는 유행이 예견되어 긴급한 예방ㆍ관리가 필요하여 질병관리청장이 보건복지부장관과 협의하여 지정하는 감염병을 포함한다.",
     ["제1급감염병”이란 생물테러감염병", "발생 또는 유행 즉시 신고하여야 하고", "음압격리와 같은 높은 수준의 격리가 필요한 감염병으로서", "질병관리청장이 보건복지부장관과 협의하여 지정하는 감염병을 포함한다"]),
    (6, 2, "제2조(정의) 제3호 · 제2급감염병",
     "“제2급감염병”이란 전파가능성을 고려하여 발생 또는 유행 시 24시간 이내에 신고하여야 하고, 격리가 필요한 다음 각 목의 감염병을 말한다. 다만, 갑작스러운 국내 유입 또는 유행이 예견되어 긴급한 예방ㆍ관리가 필요하여 질병관리청장이 보건복지부장관과 협의하여 지정하는 감염병을 포함한다.",
     ["제2급감염병”이란 전파가능성을 고려하여", "발생 또는 유행 시 24시간 이내에 신고하여야 하고", "격리가 필요한 다음 각 목의 감염병을 말한다"]),
    (7, 1, "제2조(정의) 제4호 · 제3급감염병",
     "“제3급감염병”이란 그 발생을 계속 감시할 필요가 있어 발생 또는 유행 시 24시간 이내에 신고하여야 하는 다음 각 목의 감염병을 말한다. 다만, 갑작스러운 국내 유입 또는 유행이 예견되어 긴급한 예방ㆍ관리가 필요하여 질병관리청장이 보건복지부장관과 협의하여 지정하는 감염병을 포함한다.",
     ["제3급감염병”이란 그 발생을 계속 감시할 필요가 있어", "발생 또는 유행 시 24시간 이내에 신고하여야 하는", "다음 각 목의 감염병을 말한다"]),
    (19, 3, "제13조(보건소장 등의 보고 등) ①",
     "제13조(보건소장 등의 보고 등) ① 제11조 및 제12조에 따라 신고를 받은 보건소장은 그 내용을 관할 특별자치시장ㆍ특별자치도지사 또는 시장ㆍ군수ㆍ구청장에게 보고하여야 하며, 보고를 받은 특별자치시장ㆍ특별자치도지사는 질병관리청장에게, 시장ㆍ군수ㆍ구청장은 질병관리청장 및 시ㆍ도지사에게 이를 각각 보고하여야 한다.",
     ["제11조 및 제12조에 따라 신고를 받은 보건소장은", "관할 특별자치시장ㆍ특별자치도지사 또는 시장ㆍ군수ㆍ구청장에게 보고하여야 하며", "질병관리청장에게, 시장ㆍ군수ㆍ구청장은 질병관리청장 및 시ㆍ도지사에게", "각각 보고하여야 한다"]),
]

new_blocks = []
for page, no, section, text, pieces in PLAN:
    for piece in pieces:
        ok = excerpt_in_page(piece, KEY, page)
        print(("원문 확인 " if ok else "확인 필요 "), f"PDF {page}쪽 b{no:02d} | {piece[:28]}")
        assert ok, f"{page}쪽에서 조각을 찾지 못했습니다: {piece}"
    new_blocks.append({
        "chunk_id": f"{DOC_ID}:p{page:04d}:b{no:02d}",
        "document_id": DOC_ID,
        "text": text,
        "pdf_page": page,
        "printed_page": None,
        "section": section,
    })
print("새 블록:", len(new_blocks), "개")
```

```python
ids = [b["chunk_id"] for b in new_blocks]
assert len(ids) == len(set(ids)), "새 블록 안에서 chunk_id 가 중복됩니다."
to_add = [b for b in new_blocks if b["chunk_id"] not in existing_ids]
with open(BLOCKS, "a", encoding="utf-8") as f:
    for b in to_add:
        f.write(json.dumps(b, ensure_ascii=False) + chr(10))
print("추가 저장:", len(to_add), "개 | 전체 블록:", sum(1 for _ in open(BLOCKS, encoding="utf-8")))
```
