```python
import sys
sys.path.insert(0, "../../src")
from rag.prevention_legal.load_graph import add_chunk_only
result = add_chunk_only(confirm=True)
print("적재 결과 (검사 통과):", result)
```

```python
from rag.prevention_legal.neo4j_conn import connect
d = connect()
with d.session() as s:
    print("b07 청크:", s.run("MATCH (c:Chunk {chunk_id: 'kdca_measles_2026:p0093:b07'}) RETURN c.pdf_page AS page, c.text AS text").data())
    print("전체 Chunk:", s.run("MATCH (c:Chunk) RETURN count(c) AS n").single()["n"])
    print("b07 연결:", s.run("MATCH (d:Document)-[r:HAS_CHUNK]->(c:Chunk {chunk_id: 'kdca_measles_2026:p0093:b07'}) RETURN d.document_id AS doc, count(r) AS n").data())
d.close()
```
